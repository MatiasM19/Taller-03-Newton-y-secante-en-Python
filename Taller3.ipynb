### Ejercicio 1

```python
def newton_raphson(f, df, p0, TOL=1e-6, N0=50):
    """Algoritmo 2.3. Devuelve (p, historia), con historia = [p0, p1, ..., p]."""
    historia = [p0]
    i = 1
    while i <= N0:
        d = df(p0)
        if d == 0:
            raise ZeroDivisionError(f"f'(p0) = 0 en p0 = {p0}: la tangente es horizontal")
        p = p0 - f(p0) / d        # COMPLETE: Paso 3, usando d
        historia.append(p)        # COMPLETE
        if abs(p - p0) < TOL:     # COMPLETE: Paso 4
            return p, historia
        i = i + 1
        p0 = p                    # COMPLETE: Paso 6
    raise RuntimeError(f"El método fracasó después de {N0} iteraciones")

def f(x):
    return x**3 + 4*x**2 - 10

def df(x):
    return 3*x**2 + 8*x

print(newton_raphson(f, df, 1.5))
```

### Ejercicio 2

```python
def secante(f, p0, p1, TOL=1e-6, N0=50):

    i = 2
    q0 = f(p0)
    q1 = f(p1)
    historia = [p0, p1]

    while i <= N0:

        if q1 == q0:
            raise ZeroDivisionError("Division para cero")

        p = p1 - q1 * (p1 - p0) / (q1 - q0)
        historia.append(p)

        if abs(p - p1) < TOL:
            return p, historia

        i = i + 1

        p0 = p1
        q0 = q1
        p1 = p
        q1 = f(p)

    raise RuntimeError("El método fracasó después de N0 iteraciones")

llamadas = []

def f(x):
    llamadas.append(x)
    return x**3 + 4*x**2 - 10

llamadas.clear()

p, historia = secante(f, 1.0, 2.0, TOL=1e-6)

print(p, historia, len(llamadas))
```

```python
def saldo_final(i, monto=20000.0, cuota=450.0, meses=60):
    """Saldo después de pagar todas las cuotas, con interés mensual i (0.01 es 1 %)."""
    saldo = monto
    for mes in range(meses):
        saldo = saldo * (1 + i) - cuota
    return saldo

print(saldo_final(0))
```
